```python
# Initialize cycle count globally
cycle_count = 0

# Function to add cycles to the cycle count
def add_cycles(cycles):
    global cycle_count
    cycle_count += cycles

# Initialize registers (32 registers, all set to 0)
registers = [0] * 32  # 32 registers (x0 to x31)

# Initialize memory with some values for testing
memory = [0] * 1024
memory[10] = 100  # Memory location 10 contains value 100

# Initialize program counter (pc)
pc = 0

# Initialize some registers to non-zero values for testing
registers[6] = 10  # Set x6 to 10
registers[7] = 5   # Set x7 to 5
registers[5] = 0   # Set x5 to 0 (or any value you'd like for the destination)

# Function to execute the ADD instruction
def add(dest, reg1, reg2):
    add_cycles(1)  # 1 cycle for ADD operation
    registers[dest] = registers[reg1] + registers[reg2]
    print(f"ADD: x{reg1} + x{reg2} = x{dest} -> {registers[dest]}")

# Function to execute the SUB instruction
def sub(dest, reg1, reg2):
    add_cycles(1)  # 1 cycle for SUB operation
    registers[dest] = registers[reg1] - registers[reg2]
    print(f"SUB: x{reg1} - x{reg2} = x{dest} -> {registers[dest]}")

# Function to execute the LW instruction (Load Word)
def lw(dest, address):
    add_cycles(5)  # 5 cycles for memory access (load)
    registers[dest] = memory[address]
    print(f"LW: Memory[{address}] -> x{dest} = {registers[dest]}")

# Function to execute the SW instruction (Store Word)
def sw(reg, address):
    add_cycles(5)  # 5 cycles for memory access (store)
    memory[address] = registers[reg]
    print(f"SW: x{reg} -> Memory[{address}] = {memory[address]}")

# Function to execute BEQ (Branch if Equal)
def beq(reg1, reg2, offset):
    add_cycles(1)  # 1 cycle for comparison
    if registers[reg1] == registers[reg2]:
        global pc
        pc += offset
        print(f"BEQ: x{reg1} == x{reg2}, branching to PC = {pc}")

# Function to execute BNE (Branch if Not Equal)
def bne(reg1, reg2, offset):
    add_cycles(1)  # 1 cycle for comparison
    if registers[reg1] != registers[reg2]:
        global pc
        pc += offset
        print(f"BNE: x{reg1} != x{reg2}, branching to PC = {pc}")

# Example program (a sequence of instructions)
instructions = [
    ("ADD", 5, 6, 7),  # ADD x5, x6, x7
    ("SUB", 5, 6, 7),  # SUB x5, x6, x7
    ("LW", 5, 10),     # LW x5, 10
    ("SW", 5, 20),     # SW x5, 20
    ("BEQ", 6, 7, 4),  # BEQ x6, x7, offset 4
    ("BNE", 6, 7, -4), # BNE x6, x7, offset -4
]

# Execute instructions and track cycles
for instr in instructions:
    op = instr[0]
    if op == "ADD":
        add(instr[1], instr[2], instr[3])
    elif op == "SUB":
        sub(instr[1], instr[2], instr[3])
    elif op == "LW":
        lw(instr[1], instr[2])
    elif op == "SW":
        sw(instr[1], instr[2])
    elif op == "BEQ":
        beq(instr[1], instr[2], instr[3])
    elif op == "BNE":
        bne(instr[1], instr[2], instr[3])

# After execution, print performance metrics
def print_performance_metrics(instructions):
    total_instructions = len(instructions)
    cpi = cycle_count / total_instructions
    print(f"Total Cycles: {cycle_count}")
    print(f"Total Instructions: {total_instructions}")
    print(f"CPI (Cycles Per Instruction): {cpi:.2f}")
    # Assuming a 1 GHz clock (1 billion cycles per second)
    execution_time = cycle_count / 1e9  # Time in seconds
    print(f"Execution Time: {execution_time:.6f} seconds")

# Call the performance metrics function after execution
print_performance_metrics(instructions)
```
